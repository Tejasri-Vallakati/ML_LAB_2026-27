```python
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import pandas as pd
x = np.array([1,4,6,10])
y = np.array([1,3,8,9])

plt.plot(x, y)
plt.show()
```

```python
x=[0,2,4,6,8]
y=[1,3,5,7,9]
plt.figure(figsize=(8,5))
plt.plot(x,y,color='red',linestyle='-.',marker='o',linewidth=2)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title("line chart")
plt.legend(['my data series'])
plt.grid(True)
plt.show()
```

```python
categories=['A','B','C','D','E']
values=[5,2,3,4,6]
plt.figure(figsize=(9,5))
plt.bar(categories,values,color='pink')
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Bar chart')
plt.show()
```

```python
data = [12, 15, 13, 17, 16, 16, 19, 12, 14, 16, 14, 17, 18, 18, 15, 14]
plt.hist(data, bins=16,color = 'yellow',edgecolor='blue')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Simple Histogram Example')
plt.show()
```

```python
x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
y = [5, 7, 4, 6, 8, 2, 5, 8, 7]
plt.scatter(x, y, color = 'red')
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.title('Simple Scatter Plot Example')
plt.show()
```

```python
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=0)
plt.title('Simple Pie Chart Example')
plt.axis('equal')
plt.show()
```

```python
colors = ['red', 'orange', 'yellow']  # Blue → White → Red
custom_cmap = LinearSegmentedColormap.from_list("red_orange_yellow", colors)

data = np.random.rand(10, 10)
plt.imshow(data, cmap=custom_cmap, interpolation='nearest')
plt.colorbar()
plt.title('Simple Heatmap Example')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 6, 400)
y = np.linspace(-6, 6, 400)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
Theta = np.arctan2(Y, X)

Z = np.sin(8 * Theta + R * 3)

plt.figure(figsize=(8, 8))
plt.imshow(Z, cmap='plasma', extent=(-6,6,-6,6))
plt.colorbar()
plt.title("Spiral Illusion Heatmap")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-12, 12, 1200)
y = np.linspace(-12, 12, 1200)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
A = np.arctan2(Y, X)

# Snake scales pattern
Z = np.sin(R*20 + 8*np.sin(A*6))
Z *= np.cos(A*15 + R)

plt.figure(figsize=(10,10))
plt.imshow(Z, cmap='turbo')
plt.axis('off')
plt.title("Snake Skin Vortex Illusion")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 800)
y = np.linspace(-4, 4, 800)
X, Y = np.meshgrid(x, y)

Z = np.sin(X**2 * Y) + np.cos(Y**2 * X)

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='nipy_spectral')
plt.colorbar()
plt.title("Fractal Style")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-8, 8, 700)
y = np.linspace(-8, 8, 700)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)

Z = np.sin(12 * R) * np.cos(10 * np.arctan2(Y, X))

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='turbo')
plt.colorbar()
plt.title("Vortex Illusion")
plt.axis('off')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-12, 12, 800)
y = np.linspace(-12, 12, 800)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
T = np.arctan2(Y, X)

Z = np.sin(25 * T + 6 * R)

plt.figure(figsize=(8,8))
plt.imshow(Z, cmap='twilight')
plt.colorbar()
plt.title("Hypnotic Spiral")
plt.axis('off')
plt.show()
```
